# Task 5 - Display the representation of the array data (values) and its index (index). 

Access an element of the Series by index and change its values. Assign a name to the Series object (the name property), and to its index (index name). Output its value type (dtype).

```python
scores = pd.Series(
    [72, 81, 65, 90],
    index=["A", "B", "C", "D"]
)

print("Values:")
print(scores.values)

print("\nIndex:")
print(scores.index)

print("\nValue at index B before modification:", scores["B"])
scores["B"] = 85
scores.name = "Exam Score"
scores.index.name = "Student"

print("\nUpdated Series:")
print(scores)
print("\nData type:", scores.dtype)
```

# Task 17 - Change the index of an existing DataFrame table: Add New Index Elements with `reindex()` and the `fill_value` parameter

```python
data17 = pd.DataFrame(
    {"Score": [70, 82, 91]},
    index=["A", "B", "C"]
)

new_index = pd.Index(["A", "B", "C", "D", "E"])
result17 = data17.reindex(new_index, fill_value=0)
result17
```

# Task 49 - Select all rows in the DataFrame where the value in the first column is not missing `notna()` method

```python
data49 = scores_df.copy()
data49.loc[1, "Mathematics"] = np.nan
data49.loc[3, "Mathematics"] = np.nan

result49 = data49[data49.iloc[:, 0].notna()]
result49
```

### The two Mathematics values were intentionally made missing. `data49.iloc[:, 0]` selects the first column, and `notna()` keeps rows where that column is not missing. The original `scores_df` remains unchanged because the operation uses a copy.

# Task 54 - Multiply all elements in one column of the DataFrame that are less than 100 in absolute value by 10

# Task 72 - Apply min-max normalization to one column of the DataFrame

Transform all elements `x` of the column according to the formula: `\[x' = \frac{2(x - \min(x))}{\max(x) - \min(x)} - 1\]`, where min,max are the minimum and maximum in this column.

```python
data72 = pd.DataFrame({"Score": [10, 20, 30, 40, 50]})
minimum = data72["Score"].min()
maximum = data72["Score"].max()

if maximum == minimum:
    raise ValueError("Min-max normalization is undefined when all values are equal.")

data72["Normalized"] = (
    2 * (data72["Score"] - minimum) / (maximum - minimum) - 1
)
data72
```

### The minimum value is 10 and the maximum value is 50. Therefore, the value 30 is transformed to 0. 

**The normalization maps is:**

**the minimum value to `-1`;**  
**the maximum value to `1`;**  
**values between them to corresponding positions between `-1` and `1`.**
